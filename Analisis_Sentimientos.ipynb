```python
from transformers import pipeline
import streamlit as st
import pandas as pd
import plotly.express as px
```

```python
analizador = pipeline(
    "sentiment-analysis",
    model="pysentimiento/robertuito-sentiment-analysis"
)

print("Modelo cargado correctamente")
```

```python
texto = "Me encanta esta aplicación, es excelente."

resultado = analizador(texto)

print(resultado)
```

```python
comentarios = [
    "Me encanta esta aplicación, es excelente.",
    "El servicio es terrible, no me gustó.",
    "Hoy fui a la escuela y tuve clases."
]

for texto in comentarios:
    resultado = analizador(texto)[0]

    etiqueta = resultado["label"]
    confianza = resultado["score"]

    print("Texto:", texto)
    print("Sentimiento:", etiqueta)
    print("Confianza:", confianza)
    print("-------------------------")
```

```python
datos = []

for texto in comentarios:
    resultado = analizador(texto)[0]

    datos.append({
        "Comentario": texto,
        "Sentimiento": resultado["label"],
        "Confianza": resultado["score"]
    })

df = pd.DataFrame(datos)

print(df)
```

```python
conteo = df["Sentimiento"].value_counts()

figura = px.bar(
    x=conteo.index,
    y=conteo.values,
    title="Análisis de sentimientos",
    labels={
        "x": "Tipo de sentimiento",
        "y": "Cantidad de comentarios"
    }
)

figura.show()
```
